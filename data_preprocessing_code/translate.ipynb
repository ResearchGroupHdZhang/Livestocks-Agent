```python
# 函数，实现文件夹内文件全部翻译并保存到新文件夹
# 相当于文档翻译函数
async def translate_csv(source_path,target_path):
    # source_path要翻译的文档路径
    # target_path翻译好的文档保存地址
    lines = []
    with open(source_path, "r", encoding="utf-8") as file:
        lines = file.readlines()
        
    # 只需要翻译前5行
    for i in range(5):
    # 分割每个条目
        entries = re.findall(r'\"(.*?)\"', lines[i]) # 分割出每列的内容
        for col in entries:
            trans =await chatgpt4(col)# 翻译
            lines[i] = lines[i].replace(col,trans) # 替换
    with open(target_path, "w", encoding="utf-8") as file:
        file.writelines(lines)
```
