# 完全匹配
### 与下载出来的数据data进行州县双重对应的完全匹配

```python
# 定义一个核对表头的函数(完全匹配)
def label_county(standard,data):
    data["OBJECTID"] = -1
    
    # 保留的列
    col_need = ["year","Program","Sector","Commodity","Domain","Category","Group","Data Item","class_desc","OBJECTID","State","County","Value","CV (%)"]
    data = data.loc[:,col_need]
    # 创建一个字典，将州县组合
    mapping = {(row["State"].upper(),row["County"].upper()):row["OBJECTID"] for _,row in standard.iterrows()}
    # 使用apply函数创建data的OBJECTID
    data["OBJECTID"] = data.apply(lambda row:mapping.get((row["State"],row["County"]),-1),axis=1)
    return data
```

```python
# 对data执行完全匹配
data_all = label_county(standard,data_all)
```

```python
# 对每一个大模型判断的函数
def ai_label(row,mapping):
    prompt = "美国的一个县名"+row["County"]+"，这个州"+row["State"]+"的县有："+str(mapping[row["State"]])+"，放回该县对应的所有可能的数值，只发结果"
    ai_reply = chatgpt4(prompt)
    # 在正则表达式处理
    pattern = re.compile(r"\d+")
    objectid_num = pattern.findall(ai_reply)
    if len(ai_reply)<60 and len(objectid_num)==1:
        objectid = objectid_num[0]
        print("字符串小于60："+ai_reply)
        return (objectid)
    else:
        print(ai_reply)
        objectid = -2
        # 说明有歧义不好判断
        return objectid
    
```

```python
# 模糊匹配
import re
def county_label_ai(standard,data):
    # 用嵌套字典存储州与县
    count = 0
    mapping = {}
    for _,row in standard.iterrows():
        state = row["State"].upper()
        county = row["County"].upper()
        objectid = row["OBJECTID"]
    
        if state not in mapping:
            mapping[state] = {}
        mapping[state][county] = objectid 
    while data[data["OBJECTID"]==-1].shape[0] != 0:
        print("剩余objectid为-1的行数："+str(data[data["OBJECTID"]==-1].shape[0]))
        # 只要还存在objectid=-1就执行
        objectid = ai_label(data[data["OBJECTID"]==-1].iloc[0],mapping)
        data_county = data[data["OBJECTID"]==-1].iloc[0]["County"]
        data_State = data[data["OBJECTID"]==-1].iloc[0]["State"]
        if objectid != -2:
            data.loc[(data["County"]==data_county)&(data["State"]==data_State),"OBJECTID"] = objectid
            # print("当前的"+str(data_State)+"的"+str(data_county)+"id："+str(objectid))
        else:
            data.loc[(data["County"]==data_county)&(data["State"]==data_State),"OBJECTID"] = objectid-count
            # data.loc[(data["County"]==data_county)&(data["State"]==data_State),"OBJECTID_NOTES"] = objectid[1]
            count = count+1
    return data
```
